```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import LinearRegression, Ridge, Lasso
from sklearn.neighbors import KNeighborsRegressor
from sklearn.tree import DecisionTreeRegressor
from sklearn.ensemble import AdaBoostRegressor
from sklearn.metrics import r2_score, root_mean_squared_error

df = pd.read_csv('/content/GA_5_dataset.csv')

# Separate features and target
X = df.drop(columns=['Credit_Limit'])
y = df['Credit_Limit']

# Train-test split 70:30
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# ── Q1 & Q2: Linear Regression (fit_intercept=False) ──────────────────────────
lr = LinearRegression(fit_intercept=False)
lr.fit(X_train, y_train)
y_pred_lr = lr.predict(X_test)
r2_lr = r2_score(y_test, y_pred_lr)
print(f"Q1 - Linear Regression R2 Score: {r2_lr:.3f}")
print(f"Q2 - Index of highest absolute coefficient: {np.argmax(np.abs(lr.coef_))}")

# ── Q3 & Q4: Ridge Regression ──────────────────────────────────────────────────
ridge = Ridge(solver='sag', tol=0.0005, random_state=42)
ridge.fit(X_train, y_train)
y_pred_ridge = ridge.predict(X_test)
r2_ridge = r2_score(y_test, y_pred_ridge)
print(f"\nQ3 - Ridge R2 Score: {r2_ridge:.3f}")
print(f"Q4 - Ridge intercept_: {ridge.intercept_:.3f}")

# ── Q5 & Q6: Lasso Regression ─────────────────────────────────────────────────
lasso = Lasso(alpha=100, random_state=42)
lasso.fit(X_train, y_train)
y_pred_lasso = lasso.predict(X_test)
r2_lasso = r2_score(y_test, y_pred_lasso)
coefs_in_range = np.sum((lasso.coef_ >= -1) & (lasso.coef_ <= 1))
print(f"\nQ5 - Lasso R2 Score: {r2_lasso:.3f}")
print(f"Q6 - Coefficients in range [-1, 1]: {coefs_in_range}")
print(f"     (Lasso coefficients: {lasso.coef_})")

# ── Q7: KNeighborsRegressor ────────────────────────────────────────────────────
knn = KNeighborsRegressor(n_neighbors=10, p=1)
knn.fit(X_train, y_train)
y_pred_knn = knn.predict(X_test)
rmse_knn = root_mean_squared_error(y_test, y_pred_knn)
print(f"\nQ7 - KNN RMSE: {rmse_knn:.3f}")

# ── Q8: Decision Tree Regressor ───────────────────────────────────────────────
dt = DecisionTreeRegressor(max_depth=10, min_samples_split=6, min_samples_leaf=6, random_state=42)
dt.fit(X_train, y_train)
y_pred_dt = dt.predict(X_test)
rmse_dt = root_mean_squared_error(y_test, y_pred_dt)
print(f"\nQ8 - Decision Tree RMSE: {rmse_dt:.3f}")

# ── Q9 & Q10: AdaBoost + GridSearchCV ─────────────────────────────────────────
param_grid = {
    'n_estimators': [10, 50, 100, 200, 500],
    'learning_rate': [0.1, 0.5, 1, 2]
}
ada = AdaBoostRegressor(random_state=42)
grid_search = GridSearchCV(ada, param_grid, cv=4)
grid_search.fit(X_train, y_train)

best_model = grid_search.best_estimator_
y_pred_ada = best_model.predict(X_test)
r2_ada = r2_score(y_test, y_pred_ada)
print(f"\nQ9  - AdaBoost Best Model R2 Score: {r2_ada:.3f}")
print(f"Q10 - Best n_estimators: {grid_search.best_params_['n_estimators']}")
print(f"      Best learning_rate: {grid_search.best_params_['learning_rate']}")
print(f"      Full best params: {grid_search.best_params_}")
```
