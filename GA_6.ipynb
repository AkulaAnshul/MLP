```python
from sklearn.datasets import load_wine
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import AdaBoostClassifier, BaggingClassifier, RandomForestClassifier, GradientBoostingClassifier, VotingClassifier
from sklearn.metrics import accuracy_score

wine=load_wine()
X=wine.data
y=wine.target

X_train,X_test,y_train,y_test=train_test_split(X,y,test_size=0.3,random_state=1)

lr=LogisticRegression(max_iter=10000,C=0.5,random_state=1,solver='sag')
lr.fit(X_train,y_train)
y_pred=lr.predict(X_test)
acc_lr=accuracy_score(y_test,y_pred)
print("Logistic Regression Accuracy:",round(acc_lr,3))
print("Iterations:",lr.n_iter_)

dt=DecisionTreeClassifier(random_state=1)
param_grid_dt={'criterion':['entropy','gini'],'splitter':['random','best'],'min_samples_leaf':[2,4,6,8,10],'max_depth':[3,4,5,6]}
grid_dt=GridSearchCV(dt,param_grid_dt,cv=4)
grid_dt.fit(X_train,y_train)
best_dt=grid_dt.best_estimator_
y_pred_dt=best_dt.predict(X_test)
acc_dt=accuracy_score(y_test,y_pred_dt)
print("Decision Tree Accuracy:",round(acc_dt,3))
print("Best max_depth:",grid_dt.best_params_['max_depth'])

ada=AdaBoostClassifier(random_state=1)
param_grid_ada={'n_estimators':[100,500,1000],'learning_rate':[0.5,1,2]}
grid_ada=GridSearchCV(ada,param_grid_ada,cv=4)
grid_ada.fit(X_train,y_train)
best_ada=grid_ada.best_estimator_
y_pred_ada=best_ada.predict(X_test)
acc_ada=accuracy_score(y_test,y_pred_ada)
print("AdaBoost Accuracy:",round(acc_ada,3))
print("Best n_estimators:",grid_ada.best_params_['n_estimators'])

bag=BaggingClassifier(random_state=1)
rf=RandomForestClassifier(random_state=1)
gb=GradientBoostingClassifier(random_state=1)

vote=VotingClassifier(estimators=[('bag',bag),('rf',rf),('gb',gb)],voting='hard')
vote.fit(X_train,y_train)
y_pred_vote=vote.predict(X_test)
acc_vote=accuracy_score(y_test,y_pred_vote)
print("Voting Classifier Accuracy:",round(acc_vote,3))
```
